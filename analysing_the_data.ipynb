```python

import pandas as pd
from scipy.stats import chi2_contingency
import numpy as np

# this only works datafor columns with numerical data
fd = pd.read_csv("cleaned_data_100k.csv")

def interpret (r, name1, name2):
    if r >=0.7:
        return f"STRONG POSITIVE: As {name1} increases, {name2} also strongely increases"
    elif r >=0.3:
        return f" MODERATE POSITIVE: as {name1} increasee {name2} also moderately increase"
    elif r >=-0.3:
        return f" WEAK/NO RELATION: {name1} does not affect {name2} much"
    elif r >= -0.7:
        return f"MODERATE NEGATIVE: As {name1} increases {name2} moderately decrease"
    else:
        return f" STRONG NEGATIVE: As {name1} increases {name2} decreases"

corr_age = fd['age'].corr(fd['previous_gpa'])
corr_study = fd['study_hours_per_day'].corr(fd['exam_score'])
corr_attendance = fd['attendance_percentage'].corr(fd['exam_score'])

#the two variables you can rename-eg, interp_age, interp_study
interpretation_age = interpret(corr_age,"age","previous_gpa")
interpretation_study = interpret(corr_study,"study_hours","exam_score")
interpretation_attendance = interpret(corr_attendance,"attendance_percentage","exam_score")

print(f"age vs gpa: {corr_age:.4f} {interpretation_age}")
print(f"study vs exam: {corr_study:.4f} {interpretation_study}")
print(f"attendance vs exam: {corr_attendance:.4f} {interpretation_attendance}")

#for categorical data
fd = pd.read_csv("cleaned_data_100k.csv")
fd_columns = fd.columns.str.strip().str.lower().str.replace("","_")

def cramers_v(x,y):
    confusion = pd.crosstab(x,y)
    chi2 = chi2_contingency(confusion)[0]
    n = confusion.sum().sum()
    r = min(confusion.shape) - 1
    return np.sqrt(chi2/(n*r))if r!=0 else 0
def interpret_cat(v, name1, name2):
    if v >= 0.5:
        return f"STRONG ASSOCIATION: {name1} strongly affects {name2}"
    elif v >= 0.3:
        return f"MODERATE ASSOCIATION: {name1} moderately affects {name2}"
    elif v >= 0.1:
        return f"WEAK ASSOCIATION: {name1} slightly affects{name2}"
    else:
        return f"NO ASSOCIATION {name1} does not affect {name2}"

v1 = cramers_v(fd['gender'], fd['performance_level'])
v2 = cramers_v(fd['internet_access'], fd['performance_level'])
v3 = cramers_v(fd['sleep_quality'], fd['performance_level'])
v4 = cramers_v(fd['study_consistency'], fd['performance_level'])
v5 = cramers_v(fd['parent_education'], fd['previous_gpa'])


interpretation_gender = interpret_cat(v1, "gender", "performance_level")
interpretation_internet = interpret_cat(v2, "internet access", "performance_level")
interpretation_sleep = interpret_cat(v3, "sleep quality", "performance_level")
interpretation_study = interpret_cat(v4, "study_consistency", "performance_level")
interpretation_parent = interpret_cat(v5, "parent_education", "previou_gpa")

print(f"gender vs performance_level: {v1:.4f} {interpretation_gender}")
print(f"internet vs performance_level: {v2:.4f} {interpretation_internet}")
print(f"sleep vs performance_level: {v3:.4f} {interpretation_sleep}")
print(f"study vs performance_level: {v4:.4f} {interpretation_study}")
print(f"parent_education vs previous_gpa: {v3:.4f} {interpretation_parent}")


```
