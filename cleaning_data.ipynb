```python

import pandas as pd

fd = pd.read_csv("student_exam_performance.csv")

missing = fd.isnull().sum()
missing = missing[missing >0].sort_values(ascending=False)
print("MISSING COLUMNS:","\n", missing)

numeric_cols = ["attendance_percentage", "time_management_score", "previous_gpa",]
for col in numeric_cols:
    if col in fd.columns:

        fd[col] = pd.to_numeric(fd[col], errors='coerce')
        mean_val = fd[col].mean()
        fd[col] = fd[col].fillna(mean_val)
        print(f"filled {col} with mean: {mean_val:.2f}")

categorical_cols = ["notes_quality", "parent_education", "device_availability","sleep_quality"]
for col in categorical_cols:
    if col in fd.columns:
        #mode_val = fd[col].mode()
        fd[col] = fd[col].fillna(fd[col].mode()[0])
        print(f"filled {col} with mode: {fd[col].mode()[0]}")

print("\n--- Remaining missing ---")
remaining = fd.isnull().sum()
print(remaining[remaining > 0])
print(f"\nTotal rows now: {len(fd)}")

fd.to_csv("cleaned_data_100k.csv", index=False)
print("file saved successfully as cleaned_data_100k.csv")
```
